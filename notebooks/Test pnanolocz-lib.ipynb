# Test leveling

```python
data_path = r"C:/Users/ggjh246/OneDrive - University of Leeds/Code/playNano_testdata/save-2025.05.20-12.57.06.187.h5-jpk"
# Un- comment the following line if you do not have playnano installed in your enviroment. 
#%pip install numpy pandas matplotlib playnano
from playnano.io.loader import load_afm_stack
afm_stack = load_afm_stack(data_path, channel = "height_trace")
```

```python
from matplotlib import pyplot as plt

plt.imshow(afm_stack[5], cmap="afmhot")
print(afm_stack.data.shape)
```

```python
import pnanolocz as pn

frames = afm_stack.n_frames
frame_ind = range(0, frames)
```

### Apply a quadratic plane fit

```python
plane_levelled = pn.level.apply_level(
    img=afm_stack.data, polyx=2, polyy=2, method="plane"
)
plt.imshow(plane_levelled[5], cmap="afmhot")
```

### Mask all the pixels above 0

```python
mask_hist = pn.thresholder.apply_thresholder(
    img=plane_levelled,
    method="histogram",
    limits=(float("-inf"), 0.2),
    invert=False,
)
plt.imshow(mask_hist[5], interpolation= 'none')
```

### Apply a line fit from the background (not masked) to the image

```python
masked_line_levelled = pn.level.apply_level(
    img=plane_levelled,
    polyx=1,
    polyy=0,
    method="line",
    mask=mask_hist,
)
plt.imshow(masked_line_levelled[5], cmap="afmhot")
```

### Rather than manually applying steps, use the "multi-plane-otsu" auto level routine

```python
auto_leveled = pn.level_auto.apply_level_auto(
    img_stack=afm_stack.data,
    routine="multi-plane-otsu",
)
plt.imshow(auto_leveled[5], cmap="afmhot")
```
